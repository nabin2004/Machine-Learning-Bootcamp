## 🎯 Learning Objectives
- Create and manipulate Pandas Series and DataFrames
- Load, inspect, filter, and aggregate data
- Handle missing values and merge datasets

```python
import numpy as np
import pandas as pd
print(f'Pandas version: {pd.__version__}')
```

## 1. Series and DataFrame Basics

```python
# Series – 1D labeled array
s = pd.Series([10, 20, 30, 40], index=['a', 'b', 'c', 'd'])
print('Series:'); print(s)

# DataFrame – 2D labeled table
data = {
    'Name':   ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Age':    [25, 30, 35, 28, 22],
    'Score':  [88.5, 92.0, 79.5, 95.0, 85.0],
    'Passed': [True, True, False, True, True]
}
df = pd.DataFrame(data)
print('\nDataFrame:'); df
```

## 2. Inspecting Data

```python
print('Shape:', df.shape)
print('\nInfo:'); df.info()
print('\nDescribe:'); df.describe()
```

## 3. Selecting & Filtering

```python
# Select column
print(df['Name'])

# Select multiple columns
print(df[['Name', 'Score']])

# Filter rows (boolean mask)
high_scorers = df[df['Score'] > 88]
print('High scorers:'); print(high_scorers)

# loc (label-based) and iloc (integer-based)
print('\nloc row 0-2:'); print(df.loc[0:2, ['Name', 'Age']])
print('\niloc [0:2, 0:2]:'); print(df.iloc[0:2, 0:2])
```

## 4. Adding Columns & Applying Functions

```python
df['Grade'] = df['Score'].apply(lambda s: 'A' if s >= 90 else ('B' if s >= 80 else 'C'))
df['Score_Scaled'] = (df['Score'] - df['Score'].min()) / (df['Score'].max() - df['Score'].min())
print(df)
```

## 5. Handling Missing Values

```python
df_missing = df.copy()
df_missing.loc[1, 'Score'] = np.nan
df_missing.loc[3, 'Age']   = np.nan

print('Missing values:'); print(df_missing.isnull().sum())

# Fill with mean
df_filled = df_missing.copy()
df_filled['Score'] = df_filled['Score'].fillna(df_filled['Score'].mean())
df_filled['Age']   = df_filled['Age'].fillna(df_filled['Age'].median())
print('\nAfter filling:'); print(df_filled[['Name', 'Age', 'Score']])
```

## 6. GroupBy & Aggregation

```python
print('Average score by grade:'); print(df.groupby('Grade')['Score'].mean())
print('\nCount by grade:');        print(df.groupby('Grade').size())
```
