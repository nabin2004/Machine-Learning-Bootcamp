## 🎯 Learning Objectives
- Create and manipulate NumPy arrays
- Perform vectorized operations and broadcasting
- Apply linear algebra operations needed for ML

```python
import numpy as np
print(f'NumPy version: {np.__version__}')
```

## 1. Creating Arrays

```python
# 1D array
a = np.array([1, 2, 3, 4, 5])
print('1D:', a, '| shape:', a.shape)

# 2D array (matrix)
B = np.array([[1, 2, 3],
              [4, 5, 6]])
print('2D:\n', B, '\nshape:', B.shape)

# Useful constructors
print('zeros:', np.zeros((2, 3)))
print('ones: ', np.ones((2, 3)))
print('eye:  \n', np.eye(3))
print('range:', np.arange(0, 10, 2))
print('linspace:', np.linspace(0, 1, 5))
```

## 2. Indexing & Slicing

```python
arr = np.arange(12).reshape(3, 4)
print('Array:\n', arr)
print('Row 0:         ', arr[0])
print('Column 2:      ', arr[:, 2])
print('Submatrix:\n',    arr[0:2, 1:3])
print('Boolean mask:  ', arr[arr > 6])
```

## 3. Vectorized Operations & Broadcasting

```python
x = np.array([1.0, 2.0, 3.0])
y = np.array([4.0, 5.0, 6.0])

# Element-wise (no Python loops needed!)
print('x + y =', x + y)
print('x * y =', x * y)
print('x ** 2=', x ** 2)
print('sqrt  =', np.sqrt(x))

# Dot product
print('dot(x,y) =', np.dot(x, y))

# Broadcasting: add a scalar to every element
print('x + 10 =', x + 10)
```

## 4. Aggregations & Statistics

```python
data = np.random.randn(1000)  # 1000 random standard-normal samples
print(f'mean  = {data.mean():.4f}')
print(f'std   = {data.std():.4f}')
print(f'min   = {data.min():.4f}')
print(f'max   = {data.max():.4f}')
print(f'sum   = {data.sum():.4f}')
```

## 5. Linear Algebra Essentials

```python
A = np.array([[2., 1.],
              [5., 3.]])

print('Matrix A:\n', A)
print('Transpose:\n', A.T)
print('Inverse:\n',   np.linalg.inv(A))
print('Determinant:', np.linalg.det(A))

# Solve Ax = b
b = np.array([4., 7.])
x = np.linalg.solve(A, b)
print('Solution x (Ax=b):', x)
```
