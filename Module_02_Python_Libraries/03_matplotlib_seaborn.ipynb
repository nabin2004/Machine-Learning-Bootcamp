## 🎯 Learning Objectives
- Create common plots with Matplotlib
- Build statistical visualizations with Seaborn
- Choose the right chart for the right data

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Nicer default style
sns.set_theme(style='whitegrid', palette='muted')
print(f'matplotlib: {plt.matplotlib.__version__}  |  seaborn: {sns.__version__}')
```

## 1. Matplotlib Basics

```python
x = np.linspace(0, 2 * np.pi, 200)

fig, axes = plt.subplots(1, 3, figsize=(14, 4))

# Line plot
axes[0].plot(x, np.sin(x), color='steelblue', label='sin')
axes[0].plot(x, np.cos(x), color='tomato',    label='cos')
axes[0].set_title('Line Plot'); axes[0].legend()

# Scatter plot
rng = np.random.default_rng(0)
axes[1].scatter(rng.normal(size=100), rng.normal(size=100), alpha=0.6, color='mediumpurple')
axes[1].set_title('Scatter Plot')

# Bar chart
categories = ['A', 'B', 'C', 'D']
values     = [23, 45, 12, 67]
axes[2].bar(categories, values, color='coral')
axes[2].set_title('Bar Chart')

plt.tight_layout()
plt.show()
```

## 2. Seaborn Statistical Plots

```python
# Load a built-in dataset
tips = sns.load_dataset('tips')

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Histogram + KDE
sns.histplot(tips['total_bill'], kde=True, ax=axes[0], color='steelblue')
axes[0].set_title('Distribution of Total Bill')

# Box plot
sns.boxplot(data=tips, x='day', y='total_bill', ax=axes[1])
axes[1].set_title('Total Bill by Day')

# Scatter plot coloured by category
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex', ax=axes[2])
axes[2].set_title('Tip vs Total Bill')

plt.tight_layout()
plt.show()
```

## 3. Correlation Heatmap

```python
from sklearn.datasets import load_diabetes
diabetes = load_diabetes(as_frame=True)
corr = diabetes.data.corr()

plt.figure(figsize=(9, 7))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Diabetes Dataset – Feature Correlation Heatmap')
plt.tight_layout()
plt.show()
```
