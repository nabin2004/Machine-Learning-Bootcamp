## 🎯 Learning Objectives
- Understand the K-Means algorithm step by step
- Use the elbow method and silhouette score to choose K
- Visualise and interpret clustering results

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import make_blobs

sns.set_theme(style='whitegrid')
```

## 1. The K-Means Algorithm

1. Randomly initialise K centroids.
2. Assign each point to the nearest centroid.
3. Recompute centroids as the mean of assigned points.
4. Repeat steps 2–3 until convergence.

```python
# Generate blobs with 4 true clusters
X, y_true = make_blobs(n_samples=400, centers=4, cluster_std=0.9, random_state=42)
X = StandardScaler().fit_transform(X)

kmeans = KMeans(n_clusters=4, random_state=42, n_init='auto')
labels = kmeans.fit_predict(X)

plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='tab10', alpha=0.7, s=30)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1],
            marker='X', s=200, c='black', zorder=5, label='Centroids')
plt.title('K-Means Clustering (K=4)')
plt.legend(); plt.tight_layout(); plt.show()
```

## 2. Choosing K: Elbow Method & Silhouette Score

```python
inertias, sil_scores = [], []
k_range = range(2, 11)

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init='auto')
    km.fit(X)
    inertias.append(km.inertia_)
    sil_scores.append(silhouette_score(X, km.labels_))

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

axes[0].plot(k_range, inertias, marker='o', color='steelblue')
axes[0].set_xlabel('K'); axes[0].set_ylabel('Inertia (WCSS)')
axes[0].set_title('Elbow Method')

axes[1].plot(k_range, sil_scores, marker='o', color='tomato')
axes[1].set_xlabel('K'); axes[1].set_ylabel('Silhouette Score')
axes[1].set_title('Silhouette Score (higher = better)')

plt.tight_layout(); plt.show()
best_k = k_range.start + np.argmax(sil_scores)
print(f'Best K by silhouette: {best_k}')
```

## 3. Real-World Example – Customer Segmentation

```python
import pandas as pd

np.random.seed(42)
n = 300
customers = pd.DataFrame({
    'annual_spend': np.concatenate([
        np.random.normal(5000, 800, 100),
        np.random.normal(15000, 1200, 100),
        np.random.normal(35000, 3000, 100)
    ]),
    'purchase_freq': np.concatenate([
        np.random.normal(2, 0.5, 100),
        np.random.normal(8, 1.5, 100),
        np.random.normal(20, 3, 100)
    ])
})

X_cust = StandardScaler().fit_transform(customers)
km3 = KMeans(n_clusters=3, random_state=42, n_init='auto')
customers['Segment'] = km3.fit_predict(X_cust)

plt.figure(figsize=(8, 6))
for seg in sorted(customers['Segment'].unique()):
    sub = customers[customers['Segment'] == seg]
    plt.scatter(sub['annual_spend'], sub['purchase_freq'],
                label=f'Segment {seg}', alpha=0.7, s=40)
plt.xlabel('Annual Spend ($)'); plt.ylabel('Purchase Frequency')
plt.title('Customer Segmentation with K-Means')
plt.legend(); plt.tight_layout(); plt.show()

print(customers.groupby('Segment')[['annual_spend', 'purchase_freq']].mean().round(1))
```
