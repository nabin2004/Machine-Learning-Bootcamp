## 🎯 Learning Objectives
- Understand what PCA does geometrically
- Reduce dimensionality while retaining maximum variance
- Visualise high-dimensional data in 2D and 3D

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_digits

sns.set_theme(style='whitegrid')
```

## 1. PCA Intuition

PCA finds the directions (principal components) of **maximum variance** in the data and projects the data onto a lower-dimensional subspace.

Steps:
1. Standardise the data
2. Compute the covariance matrix
3. Find eigenvectors (principal components) and eigenvalues (explained variance)
4. Project data onto the top K components

```python
# 2D example to visualise
np.random.seed(42)
mean = [0, 0]
cov  = [[3, 2], [2, 2]]
X_2d = np.random.multivariate_normal(mean, cov, 200)

pca_2d = PCA(n_components=2)
pca_2d.fit(X_2d)
components = pca_2d.components_

plt.figure(figsize=(7, 6))
plt.scatter(X_2d[:, 0], X_2d[:, 1], alpha=0.5, s=20)
origin = np.mean(X_2d, axis=0)
for comp, var, color in zip(components, pca_2d.explained_variance_, ['tomato', 'steelblue']):
    scale = np.sqrt(var) * 2
    plt.arrow(*origin, *(comp * scale), color=color, width=0.03,
              head_width=0.15, head_length=0.1, length_includes_head=True)
plt.axis('equal'); plt.title('PCA – Principal Components as Arrows')
plt.tight_layout(); plt.show()
print('Explained variance ratio:', pca_2d.explained_variance_ratio_.round(4))
```

## 2. Digits Dataset – Reducing 64 → 2 Dimensions

```python
digits = load_digits()
X_d = StandardScaler().fit_transform(digits.data)  # 1797 samples, 64 features

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_d)
print(f'Explained variance (2 PCs): {pca.explained_variance_ratio_.sum():.2%}')

plt.figure(figsize=(9, 7))
scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=digits.target, cmap='tab10', alpha=0.7, s=15)
plt.colorbar(scatter, label='Digit')
plt.title('Digits Dataset – 2D PCA Projection')
plt.xlabel('PC1'); plt.ylabel('PC2')
plt.tight_layout(); plt.show()
```

## 3. Choosing the Number of Components – Explained Variance

```python
pca_full = PCA().fit(X_d)
cumvar = np.cumsum(pca_full.explained_variance_ratio_)

n_95 = np.searchsorted(cumvar, 0.95) + 1

plt.figure(figsize=(9, 5))
plt.plot(range(1, len(cumvar) + 1), cumvar, color='steelblue')
plt.axhline(0.95, color='tomato', ls='--', label='95% variance')
plt.axvline(n_95, color='gray',   ls='--', label=f'{n_95} components')
plt.xlabel('Number of Components'); plt.ylabel('Cumulative Explained Variance')
plt.title('PCA – Choosing Number of Components')
plt.legend(); plt.tight_layout(); plt.show()
print(f'{n_95} components explain 95% of the variance')
```
