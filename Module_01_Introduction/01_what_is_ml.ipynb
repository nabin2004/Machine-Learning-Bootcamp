## 1. What is Machine Learning?

> *"Machine Learning is the field of study that gives computers the ability to learn without being explicitly programmed."*  
> — Arthur Samuel, 1959

Traditional programming:
```
Rules + Data  →  Output
```

Machine Learning:
```
Data + Output  →  Rules (Model)
```

## 3. A Tiny ML Example

Let's train our very first ML model — predicting house prices from size.

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Tiny dataset: house size (sqft) → price ($1000s)
house_size = np.array([600, 800, 1000, 1200, 1400, 1600, 1800, 2000]).reshape(-1, 1)
price      = np.array([150, 200,  250,  290,  330,  380,  410,  460])

# Train
model = LinearRegression()
model.fit(house_size, price)

# Predict
predictions = model.predict(house_size)

# Plot
plt.figure(figsize=(8, 5))
plt.scatter(house_size, price, color='steelblue', label='Actual prices', zorder=5)
plt.plot(house_size, predictions, color='tomato', linewidth=2, label='Model prediction')
plt.xlabel('House Size (sqft)')
plt.ylabel('Price ($1000s)')
plt.title('My First ML Model – House Price Prediction')
plt.legend()
plt.tight_layout()
plt.show()

print(f'Predicted price for 1500 sqft: ${model.predict([[1500]])[0]:.1f}k')
````

## 4. Key Terminology

| Term | Definition |
|------|------------|
| **Feature (X)** | Input variable (e.g., house size) |
| **Label / Target (y)** | Output variable we want to predict (e.g., price) |
| **Training set** | Data used to train the model |
| **Test set** | Held-out data used to evaluate the model |
| **Model** | The mathematical function learned from data |
| **Inference** | Using the trained model to make predictions on new data |
